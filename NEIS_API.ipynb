```python
from openai import OpenAI

base_url = "https://api.aimlapi.com/"
api_key = "7d89ea44cd584137ac387ef221d13696"
system_prompt = "You are a travel agent. Be descriptive and helpful."
user_prompt = "Tell me about San Francisco"

api = OpenAI(api_key=api_key, base_url=base_url)


def main():
    completion = api.chat.completions.create(model="mistralai/Mistral-7B-Instruct-v0.2",messages=[{"role": "system", "content": system_prompt},{"role": "user", "content": user_prompt},],temperature=0.7, max_tokens=256, )

    response = completion.choices[0].message.content

    print("User:", user_prompt)
    print("AI:", response)


main()
```
