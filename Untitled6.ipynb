```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. 가상의 데이터 생성 (Simplified VLA Embedding)
# 단어 벡터 (Query): 'Apple' (3차원 예시)
query = np.array([0.9, 0.15, -0.2])

# 영상 특징 지도 (Key): 10x10 격자 내의 사물 위치 특징값
# (5,5) 위치에 사과와 유사한 특징을 배치
keys = np.random.randn(10, 10, 3) * 0.1
keys[5, 5] = [0.85, 0.15, -0.1]  # 사과 위치 (정답 좌표)

# 2. 어텐션 스코어 계산 (Dot Product Attention)
def compute_attention(Q, K):
    # Q와 K의 내적 계산
    attention_map = np.zeros((10, 10))
    for i in range(10):
        for j in range(10):
            # 두 벡터의 내적 (Dot Product)
            attention_map[i, j] = np.dot(Q, K[i, j])

    # Softmax 시뮬레이션 (스코어 강조)
    exp_map = np.exp(attention_map * 10)
    return exp_map / np.sum(exp_map)

score_map = compute_attention(query, keys)

# 3. 시각화
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.title("Robot's Camera View (Simulated)")
view = np.zeros((10, 10))
view[5, 5] = 1 # 사과 위치 표시
sns.heatmap(view, cmap="Greens", cbar=False)
plt.xlabel("X coordinate")
plt.ylabel("Y coordinate")

plt.subplot(1, 2, 2)
plt.title("Attention Score (Mathematical Alignment)")
sns.heatmap(score_map, annot=True, fmt=".2f", cmap="YlOrRd")
plt.xlabel("Vector Space Alignment")
plt.show()

print(f"최고 어텐션 좌표: {np.unravel_index(score_map.argmax(), score_map.shape)}")
print("설명: 내적값이 가장 큰 (5,5) 좌표에서 로봇의 행동이 결정됩니다.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# 1. 환경 설정 (10x10 격자)
grid_size = 10
apple_pos = (5, 5) # 사과의 실제 위치

# 2. 데이터 초기화
# Query: 'Apple' 명령어 벡터
query = np.array([1.0, 0.5])
# Keys: 공간 특징 벡터 (사과 위치만 Query와 유사하게 설정)
keys = np.random.randn(grid_size, grid_size, 2) * 0.1
keys[apple_pos] = [0.95, 0.45]

fig, ax = plt.subplots(figsize=(6, 5))
plt.title("VLA Model: Attention Convergence (Softmax/Dot-product)")

# 애니메이션 업데이트 함수
def update(frame):
    ax.clear()
    # 프레임이 지날수록 노이즈가 줄어들고 수학적 정렬(Alignment)이 강화됨을 시각화
    strength = min(frame / 20.0, 10.0)

    # 내적(Dot Product) 계산
    attention_map = np.zeros((grid_size, grid_size))
    for i in range(grid_size):
        for j in range(grid_size):
            attention_map[i, j] = np.dot(query, keys[i, j])

    # Softmax 연산 시뮬레이션
    exp_map = np.exp(attention_map * strength)
    score_map = exp_map / np.sum(exp_map)

    im = ax.imshow(score_map, cmap='YlOrRd', interpolation='gaussian')
    ax.set_title(f"Frame {frame}: Mathematical Alignment Strength {strength:.1f}")
    return [im]

# 애니메이션 생성 (50프레임)
ani = FuncAnimation(fig, update, frames=50, interval=100, blit=True)

# 코랩에서 영상을 재생하기 위한 처리
plt.close()
HTML(ani.to_jshtml())
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# 1. 환경 설정
grid_size = 10
apple_pos = np.array([7, 7])  # 사과 위치 (Target)
arm_start_pos = np.array([1, 1])  # 로봇 팔 시작 위치

# 2. 임베딩 벡터 설정 (수학적 모델)
query_text = np.array([1.0, 0.8])  # "사과를 집어라" 명령 벡터
apple_key = np.array([0.95, 0.75]) # 사과의 특징 벡터 (Query와 유사)
other_keys = np.random.randn(grid_size, grid_size, 2) * 0.1 # 배경 노이즈

# 3. 그래프 초기화
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

def update(frame):
    ax1.clear()
    ax2.clear()

    # 정렬 강도 (시간이 지날수록 수학적 확신이 강해짐)
    strength = min(frame / 10.0, 15.0)

    # --- 오른쪽: 어텐션 맵 연산 (수학적 공간) ---
    attention_map = np.zeros((grid_size, grid_size))
    for i in range(grid_size):
        for j in range(grid_size):
            # 실제 사과 위치에만 apple_key 부여
            k = apple_key if (i, j) == tuple(apple_pos) else other_keys[i, j]
            attention_map[i, j] = np.dot(query_text, k)

    # Softmax 시뮬레이션
    exp_map = np.exp(attention_map * strength)
    score_map = exp_map / np.sum(exp_map)

    im2 = ax2.imshow(score_map, cmap='YlOrRd', interpolation='gaussian')
    ax2.set_title(f"Right: Attention Score (Math)\nAlignment Strength: {strength:.1f}")

    # --- 왼쪽: 로봇 시점 (물리적 공간) ---
    # 어텐션이 가장 높은 곳으로 로봇 팔이 이동 (간단한 보간)
    target_idx = np.unravel_index(score_map.argmax(), score_map.shape)
    progress = min(frame / 50.0, 1.0)
    current_arm_pos = arm_start_pos + (np.array(target_idx) - arm_start_pos) * progress

    ax1.set_xlim(-0.5, grid_size-0.5)
    ax1.set_ylim(-0.5, grid_size-0.5)
    ax1.grid(True, linestyle='--', alpha=0.5)

    # 사과(목표) 표시
    ax1.scatter(apple_pos[1], apple_pos[0], c='red', s=200, label='Target (Apple)', marker='o')
    # 로봇 팔(현재 위치) 표시
    ax1.scatter(current_arm_pos[1], current_arm_pos[0], c='blue', s=100, label='Robot Arm', marker='s')
    # 팔의 이동 경로 선
    ax1.plot([arm_start_pos[1], current_arm_pos[1]], [arm_start_pos[0], current_arm_pos[0]], 'b--', alpha=0.3)

    ax1.set_title(f"Left: Robot View (Physical)\nFrame: {frame}")
    ax1.legend(loc='upper left')
    ax1.invert_yaxis() # 행렬 인덱스와 시각적 좌표 일치

    return [im2]

# 애니메이션 생성
ani = FuncAnimation(fig, update, frames=50, interval=150, blit=False)
plt.close()
HTML(ani.to_jshtml())
```
