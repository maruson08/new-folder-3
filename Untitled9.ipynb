```python
import cv2
import face_recognition
import hashlib
import numpy as np

def extract_face_embedding(image):
    # BGR → RGB
    rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # 얼굴 탐지
    face_locations = face_recognition.face_locations(rgb)

    if len(face_locations) == 0:
        raise ValueError("얼굴을 찾을 수 없음")

    # embedding 추출
    face_encodings = face_recognition.face_encodings(rgb, face_locations)

    return face_encodings[0]


def embedding_to_hash(embedding):
    # numpy → bytes → hash
    embedding_bytes = embedding.tobytes()
    return hashlib.sha256(embedding_bytes).hexdigest()


def process_image(image_path):
    image = cv2.imread(image_path)

    if image is None:
        raise ValueError("이미지 로드 실패")

    embedding = extract_face_embedding(image)
    hash_value = embedding_to_hash(embedding)

    return hash_value


# 실행
try:
    result_hash = process_image(image_path)
    print("얼굴 해시:", result_hash)
except Exception as e:
    print("오류:", e)
```

# colab js based live webcam

```python
from IPython.display import display, Javascript
from google.colab.output import eval_js
import base64
import numpy as np
import cv2

def get_frame():
    js = Javascript('''
    async function captureFrame() {
        const video = document.createElement('video');
        const stream = await navigator.mediaDevices.getUserMedia({video: true});
        video.srcObject = stream;
        await video.play();

        const canvas = document.createElement('canvas');
        canvas.width = video.videoWidth;
        canvas.height = video.videoHeight;

        const context = canvas.getContext('2d');
        context.drawImage(video, 0, 0);

        stream.getTracks().forEach(track => track.stop());

        return canvas.toDataURL('image/jpeg');
    }
    captureFrame();
    ''')
    display(js)

    data = eval_js("captureFrame()")
    binary = base64.b64decode(data.split(',')[1])

    np_arr = np.frombuffer(binary, np.uint8)
    frame = cv2.imdecode(np_arr, cv2.IMREAD_COLOR)

    return frame
```

```python
import face_recognition
import hashlib

def extract_face_embedding(image):
    rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    face_locations = face_recognition.face_locations(rgb)

    if len(face_locations) == 0:
        return None, None

    encodings = face_recognition.face_encodings(rgb, face_locations)

    return encodings[0], face_locations[0]


def embedding_to_hash(embedding):
    return hashlib.sha256(embedding.tobytes()).hexdigest()
```

```python
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time

while True:
    frame = get_frame()

    embedding, face_loc = extract_face_embedding(frame)

    if embedding is not None:
        hash_value = embedding_to_hash(embedding)

        top, right, bottom, left = face_loc
        cv2.rectangle(frame, (left, top), (right, bottom), (0,255,0), 2)

        cv2.putText(frame, hash_value[:10],
                    (left, top-10),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.6, (0,255,0), 2)

    # 화면 출력
    clear_output(wait=True)
    plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
    plt.axis("off")
    plt.show()

    time.sleep(1)  # 속도 조절 (1초마다 캡처)
```
